```python
import pandas as pd
import numpy as np

# 1. Necessary cleaned datasets load karein
orders_df = pd.read_csv('cleaned_orders.csv')
order_items_df = pd.read_csv('cleaned_order_items.csv')
customers_df = pd.read_csv('cleaned_customers.csv')

# 2. Datetime columns ko convert karein
orders_df['order_purchase_timestamp'] = pd.to_datetime(orders_df['order_purchase_timestamp'])

# 3. Sirf delivered orders ko filter karein aur item values ko merge karein
delivered_orders = orders_df[orders_df['order_status'] == 'delivered']
order_revenue = order_items_df.groupby('order_id').agg({'price': 'sum', 'freight_value': 'sum'}).reset_index()
order_revenue['total_order_value'] = order_revenue['price'] + order_revenue['freight_value']

# 4. Customers, Orders, aur Revenue ko merge karna (customer_unique_id ke base par taaki repeat buyers track ho sakein)
full_df = delivered_orders.merge(customers_df, on='customer_id', how='inner')
full_df = full_df.merge(order_revenue, on='order_id', how='inner')

# Snapshot date set karein (Dataset ki max purchase date se 1 din aage)
snapshot_date = full_df['order_purchase_timestamp'].max() + pd.Timedelta(days=1)

# 5. RFM Metrics Calculate karna per customer_unique_id
rfm_df = full_df.groupby('customer_unique_id').agg({
    'order_purchase_timestamp': lambda x: (snapshot_date - x.max()).days, # Recency
    'order_id': 'nunique',                                                # Frequency
    'total_order_value': 'sum'                                            # Monetary
}).reset_index()

# Column names rename karein
rfm_df.columns = ['customer_unique_id', 'Recency', 'Frequency', 'Monetary']

# 6. RFM Scores (1 to 5 scale using Quantiles) assign karna
# Recency mein kam din achhe hote hain, isliye labels reverse hain (5 = sabse recent)
rfm_df['R_Score'] = pd.qcut(rfm_df['Recency'], q=5, labels=[5, 4, 3, 2, 1]).astype(int)
# Frequency aur Monetary mein zyada value achhi hoti hai (5 = sabse zyada)
rfm_df['F_Score'] = pd.qcut(rfm_df['Frequency'].rank(method='first'), q=5, labels=[1, 2, 3, 4, 5]).astype(int)
rfm_df['M_Score'] = pd.qcut(rfm_df['Monetary'], q=5, labels=[1, 2, 3, 4, 5]).astype(int)

# Total RFM Score combine karna
rfm_df['RFM_Score'] = rfm_df['R_Score'].astype(str) + rfm_df['F_Score'].astype(str) + rfm_df['M_Score'].astype(str)

# 7. Customer Segments Define karna (Business Logic)
def rfm_segmenter(df):
    if df['R_Score'] >= 4 and df['F_Score'] >= 4:
        return 'Champions'
    elif df['R_Score'] >= 3 and df['F_Score'] >= 3:
        return 'Loyal Customers'
    elif df['R_Score'] >= 4 and df['F_Score'] == 1:
        return 'Recent Customers'
    elif df['R_Score'] <= 2 and df['F_Score'] >= 3:
        return 'At Risk'
    elif df['R_Score'] <= 2 and df['F_Score'] <= 2:
        return 'Hibernating / Lost'
    else:
        return 'Potential Loyalists'

rfm_df['Customer_Segment'] = rfm_df.apply(rfm_segmenter, axis=1)

# 8. Result check karein aur CSV save karein
print(rfm_df['Customer_Segment'].value_counts())
rfm_df.to_csv('cleaned_customer_rfm.csv', index=False)
print("\nRFM Analysis completed and saved as 'cleaned_customer_rfm.csv'!")
```

```python

from sqlalchemy import create_engine
import pandas as pd

# Step 1: PostgreSQL Connection Details
username = "postgres"
password = "Aish123"
host = "localhost"
port = "5432"
database = "olist_db"

# Step 2: Create PostgreSQL Engine
engine = create_engine(
    f"postgresql+psycopg2://{username}:{password}@{host}:{port}/{database}"
)

# Step 3: Test Connection
with engine.connect() as connection:
    print("PostgreSQL connection successful!")

# Step 4: DataFrames Dictionary
tables_to_export = {
    'customers': Customers,
    'geolocation': Geolocation,
    'order_items': Order_items,
    'order_payments': Order_payments,
    'order_reviews': Order_reviews,
    'orders': Orders,
    'products': Products,
    'sellers': Sellers,
    'product_category_name_translation': Product_category_name_translation,
    'rfm': rfm_df
}

# Step 5: Export DataFrames
for name, df in tables_to_export.items():

    df.to_sql(
        name=name,
        con=engine,
        if_exists='replace',
        index=False,
        chunksize=10000
    )

    print(f"Table '{name}' successfully exported - {len(df)} rows")

# Step 6: Final Message
print("\nAll tables successfully pushed to PostgreSQL!")

engine.dispose()
```
